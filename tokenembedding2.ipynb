```python
import torch
input_ids = torch.tensor([2,3,5,1])
```

we are doing embedding here we will get every vector embedding of dimension 3 and vocabulary size for this embedding model will be 8

```python
voacb_size = 8
output_dim = 3
torch.manual_seed(123)
embedding_layer = torch.nn.Embedding(voacb_size,output_dim)
```

```python
print(embedding_layer.weight)
```

get the vector for the given id as it is staring from 0th index


```python
print(embedding_layer(torch.tensor(3)))
```

```python
print(embedding_layer(input_ids))
```
